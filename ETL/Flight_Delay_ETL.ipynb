#MEMBUAT FACT TABLE

Fact table berisi data utama transaksi penerbangan seperti:

- tahun
- bulan
- maskapai
- bandara asal
- delay keberangkatan
- delay kedatangan

```python
fact_flights = flights[
[
'YEAR',
'MONTH',
'DAY',
'AIRLINE',
'ORIGIN_AIRPORT',
'DESTINATION_AIRPORT',
'DEPARTURE_DELAY',
'ARRIVAL_DELAY',
'CANCELLED'
]
]
```

#VISUALISASI 1

Jenis Visualisasi : Bar Chart

Tujuan Visualisasi : Visualisasi ini digunakan untuk melihat maskapai dengan rata-rata keterlambatan kedatangan tertinggi.

penjelasan :
Data dikelompokkan berdasarkan kode maskapai (AIRLINE), kemudian dihitung rata-rata keterlambatan kedatangannya (ARRIVAL_DELAY).

Grafik batang digunakan karena lebih mudah membandingkan nilai delay antar maskapai.

Semakin tinggi batang:
semakin besar rata-rata keterlambatan maskapai tersebut.

```python
avg_delay_airline = fact_flights.groupby('AIRLINE')['ARRIVAL_DELAY'].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))

avg_delay_airline.plot(kind='bar')

plt.title('Top 10 Airline with Highest Arrival Delay')
plt.xlabel('Airline')
plt.ylabel('Average Arrival Delay')

plt.show()
```

#VISUALISASI 2

Jenis Visualisasi : Horizontal Bar Chart

Tujuan Visualisasi : Visualisasi ini digunakan untuk melihat bandara dengan rata-rata keterlambatan kedatangan tertinggi.

Penjelasan : Data dikelompokkan berdasarkan kode bandara asal (ORIGIN_AIRPORT) lalu dihitung rata-rata delay kedatangannya.

Grafik horizontal dipilih agar nama atau kode bandara lebih mudah dibaca.

Semakin panjang batang:
semakin tinggi rata-rata keterlambatan pada bandara tersebut.

```python
airport_delay = fact_flights.groupby('ORIGIN_AIRPORT')['ARRIVAL_DELAY'].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))

airport_delay.plot(kind='barh')

plt.title('Top 10 Airport with Highest Arrival Delay')
plt.xlabel('Average Arrival Delay')
plt.ylabel('Airport')

plt.show()
```

# VISUALISASI 3

Jenis Visualisasi : Histogram

Tujuan Visualisasi : Visualisasi ini digunakan untuk melihat distribusi atau penyebaran data keterlambatan kedatangan pesawat.

Penjelasan :

Histogram menunjukkan jumlah penerbangan pada rentang delay tertentu.

Data dibatasi dari:

lebih besar dari -50
lebih kecil dari 200

agar data ekstrem tidak membuat grafik sulit dibaca.

Semakin tinggi batang histogram:
➡ semakin banyak jumlah penerbangan pada rentang delay tersebut.

```python
filtered_delay = fact_flights[
(fact_flights['ARRIVAL_DELAY'] > -50) &
(fact_flights['ARRIVAL_DELAY'] < 200)
]

plt.figure(figsize=(10,5))

filtered_delay['ARRIVAL_DELAY'].plot(kind='hist', bins=30)

plt.title('Distribution of Arrival Delay')
plt.xlabel('Arrival Delay')
plt.ylabel('Frequency')

plt.show()
```

#VISUALISASI 4

Jenis Visualisasi : Box Plot

Tujuan Visualisasi : Visualisasi ini digunakan untuk melihat persebaran data delay serta mendeteksi adanya outlier atau delay ekstrem.

Box plot membantu melihat:

nilai tengah data (median)
persebaran data
data ekstrem atau outlier

Titik yang jauh dari kotak menunjukkan adanya penerbangan dengan delay yang sangat tinggi.

```python
plt.figure(figsize=(8,5))

plt.boxplot(filtered_delay['ARRIVAL_DELAY'])

plt.title('Box Plot of Arrival Delay')
plt.ylabel('Arrival Delay')

plt.show()
```

#VISUALISASI 5

Jenis Visualisasi : Scatter Plot

Tujuan Visualisasi : Visualisasi ini digunakan untuk melihat hubungan antara keterlambatan keberangkatan dan keterlambatan kedatangan.

Penjelasan :

Setiap titik pada grafik mewakili satu penerbangan.

Sumbu X : delay keberangkatan

Sumbu Y : delay kedatangan

Jika titik cenderung naik ke kanan:
penerbangan yang terlambat berangkat biasanya juga terlambat tiba.

Digunakan sample(500) agar grafik tidak terlalu padat karena dataset sangat besar.

```python
sample_data = fact_flights.sample(500)

plt.figure(figsize=(10,5))

plt.scatter(
sample_data['DEPARTURE_DELAY'],
sample_data['ARRIVAL_DELAY']
)

plt.title('Departure Delay vs Arrival Delay')
plt.xlabel('Departure Delay')
plt.ylabel('Arrival Delay')

plt.show()
```

#VISUALISASI 6

Jenis Visualisasi : Area Chart

Tujuan Visualisasi : Visualisasi ini digunakan untuk melihat perbandingan rata-rata delay antar maskapai dengan tampilan area.

Penjelasan :

Area chart mirip seperti line chart, tetapi area di bawah garis diisi warna sehingga perubahan data lebih mudah terlihat.

Semakin tinggi area:
semakin tinggi rata-rata keterlambatan maskapai tersebut.

```python
airline_delay = fact_flights.groupby('AIRLINE')['ARRIVAL_DELAY'].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))

airline_delay.plot(kind='area')

plt.title('Average Arrival Delay per Airline')
plt.xlabel('Airline')
plt.ylabel('Average Arrival Delay')

plt.show()
```
